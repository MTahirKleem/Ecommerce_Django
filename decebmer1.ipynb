```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data.csv')
```

```python
df.head()
```

```python
df.Category.unique()
```

```python
# Create a new column named Categories_id and assign a numeric value to each Category column value 1 for Westernwear-Women, 2 for Indianwear-Women, 3 for Lingere, 4 for Footwear and 5 for Watches
df.loc[df['Category'] == 'Westernwear-Women', 'Category'] = 1
df.loc[df['Category'] == 'Indianwear-Women', 'Category'] = 2
df.loc[df['Category'] == 'Lingerie&Nightwear-Women', 'Category'] = 3
df.loc[df['Category'] == 'Footwear-Women', 'Category'] = 4
df.loc[df['Category'] == 'Watches-Women', 'Category'] = 5


```

```python
# rename the column name from Category to Categories_id
df.rename(columns={'Category': 'Categories_id'}, inplace=True)
```

```python
df.head(2)
```

```python
df['featured_image'] = 'media/featured_image/default.jpg'
```

```python
df['id'] = np.arange(1, len(df) + 1)
```

```python
df.info()
```

```python
final_df = df[['id', 'total_quantity', 'Availability',  'product_name', 'price', 'Discount', 'model_Name', 'Description', 'Categories_id', 'slug', 'shipping_fee', 'featured_image']]
```
